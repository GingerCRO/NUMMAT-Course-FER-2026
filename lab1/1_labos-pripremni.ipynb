**3a)** U računalu rješavamo kvadratnu jednadžbu
$$(x+10^{7})(x-10^{-7})=0, \quad\text{odnosno}\quad x^2+(10^{7}-10^{-7})x-1=0.$$
Odredite oba rješenja pomoću formule
$$x_{1,2}=\frac{-b\pm\sqrt{b^2-4ac}}{2a}$$
te za svako izračunajte relativnu grešku u odnosu na egzaktna rješenja $x_1 = 10^{-7}$ i $x_2 = -10^{7}$.

```python
# Ovdje ide rješenje (Rješenje je kod ako se u zadatku traži kodiranje i objašnjenje u Markdown ćeliji ako se traži objašnjenje)
import math

def rezultati_kvadratna(a, b, c):
    d = (b**2) - (4 * a * c)
    diskriminanta = math.sqrt(d)

    x1 = (-b - diskriminanta) / (2 * a)
    x2 = (-b + diskriminanta) / (2 * a)

    return x1, x2

a = 1
b = 10**7 - 10**(-7)
c = -1

x1, x2 = rezultati_kvadratna(a, b, c)

x1_egzaktno = 10**(-7)
x2_egzaktno = -10**7

relativna_greska_x1 = abs(x1_egzaktno - x1) / abs(x1_egzaktno)
relativna_greska_x2 = abs(x2_egzaktno - x2) / abs(x2_egzaktno)

print(f"Egzaktni x1 = {x1_egzaktno}")
print(f"Dobiveni x1 = {x1}")
print(f"Relativna greška za x1 = {relativna_greska_x1}")
print()
print(f"Egzaktni x2 = {x2_egzaktno}")
print(f"Dobiveni x2 = {x2}")
print(f"Relativna greška za x2 = {relativna_greska_x2}")
```

Zašto smo dobili samo jedno rješenje, iako sustav ima beskonačno mnogo rješenja? Zadovoljava li dobiveno rješenje prvu jednadžbu?

```python
# Ovdje ide rješenje (Rješenje je kod ako se u zadatku traži kodiranje i objašnjenje u Markdown ćeliji ako se traži objašnjenje)

# Definiramo matrice A i b
A = np.array([[1.0, 7.0], [0.1, 0.7]])
b = np.array([3.0, 0.3])

# Riješi linearni sustav
rjesenje_sustava = np.linalg.solve(A, b)
print(f"Rješenja sustava: x = {rjesenje_sustava[0]}, y = {rjesenje_sustava[1]}")

# Determinanta i uvjetovanost
determinanta = np.linalg.det(A)
uvjetovanost = np.linalg.cond(A)
print(f"det(A) = {determinanta}")
print(f"cond(A) = {uvjetovanost}")

# Ispis brojeva 0.7 i 7 * 0.1 sa 17 značajnih znamenaka
print(f"{0.7:.16f}")
print(f"{7 * 0.1:0.16f}")

# Zadovoljava li dobiveno rješenje prvu jednadžbu
uvjet = (rjesenje_sustava[0] + 7 * rjesenje_sustava[1]) == 3
print(f"{rjesenje_sustava[0]} + 7 * {rjesenje_sustava[1]} = 3 --> {uvjet}")

# Dobili smo jedno rješenje, jer se brojevi poput 0.1 i 0.7 ne mogu skroz točno zapisati u memoriji računala.
# 7 * 0.1 se zapravo u memoriji računala sprema kao 0.700000000000001
# Zbog toga je determinanta različita od nule, što znači da ima jedistveno rješenje.
```

**5a)** Odredite LU faktorizaciju **bez pivotiranja** matrice
$$A=\begin{bmatrix}10^{-17} & 1\\ 1 & 1\end{bmatrix},\qquad b=\begin{bmatrix}1\\2\end{bmatrix}$$
koristeći funkcije `LU_dekompozicija`, `forward` i `backward` iz [bilježnice s 3. vježbe](https://anazga.github.io/NumMat_Lab/Labos03.html) (malo izmijenjene verzije nalaze se u sljedećoj ćeliji). Ispišite matrice $L$ i $U$ te umnožak $LU$ i usporedite ga s matricom $A$. Riješite sustav $Ax=b$ tako da funkcijama `forward` i `backward` riješite trokutaste sustave $Ly=b$, $Ux=y$, te usporedite dobiveno rješenje s rješenjem funkcije `np.linalg.solve`. Izračunajte uvjetovanost matrice $A$ (`np.linalg.cond`).

Odgovorite u Markdown ćeliji: je li uzrok pogreške loša uvjetovanost matrice ili algoritam? Koji element matrice $U$ je izračunat netočno i kakve to veze ima sa zadatkom 1b?

```python
# ovo su funkcije iz 3. laboratorijske vježbe

def LU_dekompozicija(A):
    A = np.array(A, dtype=float)   # kopija u floatu, original ostaje netaknut
    n = A.shape[0]
    for k in range(n-1):
        for i in range(k+1, n):
            A[i,k] = A[i,k]/(A[k,k])   # multiplikator
            A[i, k+1:n] = A[i, k+1:n] - A[i,k]*A[k, k+1:n]
    return A

#A = LU_dekompozicija(A)
#Za prikaz L upotrijebiti np.tril(A, -1)+np.identity(len(A)), za U np.triu(A) 

def forward(L,b):
# supstitucija unaprijed
    b = np.array(b, dtype=float)
    b[0] = b[0] / L[0, 0]
    n=len(L)
    for i in range(1, n):
        b[i] = (b[i] - np.dot(L[i,:i], b[:i])) / L[i,i]
    return b

def backward(U,b):
# supstitucija unatrag
    b = np.array(b, dtype=float)
    n=len(U)
    for k in range(n-1,-1,-1):
        b[k] = (b[k] - np.dot(U[k,k+1:n],b[k+1:n]))/U[k,k]
    return b

# Ovdje ide rješenje (Rješenje je kod ako se u zadatku traži kodiranje i objašnjenje u Markdown ćeliji ako se traži objašnjenje)

# Definiranje sustava
A = np.array([[10**(-17), 1.0], [1.0, 1.0]])
b = np.array([1.0, 2.0])

# LU dekompozicija
A_lu = LU_dekompozicija(A)

# Dohvaćanje L i U matrica iz A_lu
L = np.tril(A_lu, -1) + np.identity(len(A_lu))
U = np.triu(A_lu)
print("Matrica L:\n", L)
print()
print("Matrica U:\n", U)

# Umnožak L * U
LU_umnozak = np.dot(L, U)
print("\nUmnožak L * U:\n", LU_umnozak)
print()
print("Originalna matrica A:\n", A)

# RJEŠAVANJE SUSTAVA Ax = b

# Prvi korak: Ly = b
y = forward(L, b)

# Drugi korak: Ux = y
x_lu = backward(U, y)
print(f"\nRješenja preko LU dekompozicije: x = {x_lu[0]}, y = {x_lu[1]}")

# Rješenje sustava preko np.linalg.solve
x_linalg_solve = np.linalg.solve(A, b)
print(f"\nRješenja preko np.linalg.solve: x = {x_linalg_solve[0]}, y = {x_linalg_solve[1]}\n")

# Uvjetovanost matrice A
uvjetovanost = np.linalg.cond(A)
print(f"cond(A) = {uvjetovanost}")

# Pogreška se dogodila zbog nedostatka pivotiranja. Dobivena uvjetovanost od 2.61 nam govori da je matrica A dobro uvjetovana.
# Algoritam bez pivotiranja je dijelio s jako malim elementom, u našem slučaju je to 10^(-17).

# U drugom retku računamo 1 - L21 * U12, odnosno 1 - 10^17 * 1 = 1 - 10^17 --> rezultat će biti otprilike -10^17 (jedinica se "zanemaruje")
```
